```python
import pandas as pd

df = pd.read_csv("student_lifestyle_dataset.csv")
df = df.rename(columns={"Stress_Level": "Burnout_Level"})

```

```python
import matplotlib.pyplot as plt

df["Burnout_Level"].value_counts().plot(kind="bar")
plt.title("Burnout Level Distribution")
plt.xlabel("Burnout Level")
plt.ylabel("Number of Students")
plt.show()
```

```python
df.boxplot(column="Study_Hours_Per_Day", by="Burnout_Level")
plt.title("Study Hours vs Burnout Level")
plt.suptitle("")
plt.ylabel("Study Hours Per Day")
plt.show()
```

```python
df.boxplot(column="Sleep_Hours_Per_Day", by="Burnout_Level")
plt.title("Sleep Hours vs Burnout Level")
plt.suptitle("")
plt.ylabel("Sleep Hours Per Day")
plt.show()
```

```python
df.boxplot(column="Physical_Activity_Hours_Per_Day", by="Burnout_Level")
plt.title("Physical Activity vs Burnout Level")
plt.suptitle("")
plt.ylabel("Hours Per Day")
plt.show()
```

```python
df.boxplot(column="Social_Hours_Per_Day", by="Burnout_Level")
plt.title("Social Time vs Burnout Level")
plt.suptitle("")
plt.ylabel("Hours Per Day")
plt.show()
```

```python
burnout_map = {"Low": 0, "Moderate": 1, "High": 2}
df["Burnout_Score"] = df["Burnout_Level"].map(burnout_map)
```

```python
df.corr(numeric_only=True)["Burnout_Score"].sort_values(ascending=False)
```

```python
def burnout_recommendations(row):
    tips = []

    if row.Sleep_Hours_Per_Day < 7:
        tips.append("Increase sleep to at least 7–8 hours per night.")

    if row.Study_Hours_Per_Day > 6:
        tips.append("Reduce daily study hours or divide study sessions.")

    if row.Physical_Activity_Hours_Per_Day < 1:
        tips.append("Increase physical activity to reduce mental fatigue.")

    if row.Social_Hours_Per_Day < 1:
        tips.append("Spend more time socializing to balance academic stress.")

    if row.Extracurricular_Hours_Per_Day > 3:
        tips.append("Limit extracurricular workload to avoid overload.")

    return tips

df["Recommendations"] = df.apply(burnout_recommendations, axis=1)
```

```python
from collections import Counter

high_burnout = df[df["Burnout_Level"] == "High"]

all_tips = []
for tips in high_burnout["Recommendations"]:
    all_tips.extend(tips)

Counter(all_tips)
```
